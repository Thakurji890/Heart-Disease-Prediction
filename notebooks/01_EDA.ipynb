```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use("ggplot")
```

## Load Dataset

```python
df = pd.read_csv("../data/raw/heart2.csv")
df.head()
```

## Missing Values

```python
df.isnull().sum()
```

## Duplicate Rows

```python
duplicates = df.duplicated().sum()
print(duplicates)
```

## Target Distribution  
# checking for balanced or not

```python
sns.countplot(data=df,x="target")
plt.show()
```

## Distribution of Numerical Features

```python
df.hist(figsize=(15,12))
plt.show()
```

## Correlation Matrix

```python
plt.figure(figsize=(12,10))
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap="coolwarm"
)
plt.show()
```
